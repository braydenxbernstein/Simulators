## 1. Bottom Line Up Front

> **We propose building *Findable*, a subscription web app that shows local businesses whether AI assistants like ChatGPT recommend them, who gets recommended instead, and exactly what to fix.**

- **The problem:** Customers are shifting from Google to AI chatbots to find local businesses, but owners have no way to see how they show up in those answers. Most are invisible and don't know it.
- **The solution:** Findable uses the OpenAI API to run hundreds of realistic customer questions ("best brake shop in Tempe," "dog groomer open Sundays near me"), parse the answers into structured data, and turn the results into a visibility score and a prioritized action plan.
- **Why we should care:** This is a fast-growing, mostly unserved market. Traditional SEO tools measure Google rankings, not AI recommendations. A low-cost, self-serve tool lets us capture small businesses early and upsell agencies managing many clients.

## 2. Business Problem

### Who experiences it?
| Stakeholder | Pain point |
|---|---|
| **Local business owners** (auto shops, restaurants, salons, dentists, contractors) | Don't know if ChatGPT recommends them or their competitors |
| **Marketing agencies** serving local clients | Clients are asking about "AI SEO" and agencies have no reporting tool to prove results |
| **Multi-location franchises** | Need to monitor visibility consistently across dozens of cities |

### Why is it worth solving?
- **Consumer behavior has shifted fast.** BrightLocal's 2026 Local Consumer Review Survey found that **45% of consumers now use AI tools for local business recommendations, up from 6% the year before**, making AI the third most popular source of recommendations. ChatGPT alone was used by **31%** of consumers for a business recommendation in the past year.
- **Trust is high.** In the same research, **64% of AI users** said they trust tools like ChatGPT as much as reviews when choosing a local business.
- **Most businesses are invisible.** A Local Falcon analysis of roughly 190,000 ChatGPT results found **83% of restaurants never appeared in ChatGPT answers**, compared to only 14% being invisible on Google.
- **The trend is accelerating.** Gartner predicted traditional search engine volume would drop 25% by 2026 as AI chatbots become "substitute answer engines." Yext's 2026 survey of 3,848 consumers found 44% used AI for local search in the past month, and among top US earners AI has already overtaken Google as the starting point.

**Bottom line:** A small business can be #1 on Google Maps and still lose customers every day to a competitor that ChatGPT recommends instead. Right now they have no visibility into that channel.

## 3. Proposed Solution

**Findable** is a web dashboard that answers three questions for a business owner:

1. **"Am I being recommended?"** A visibility score (0 to 100) based on how often the business appears in AI answers across realistic customer prompts.
2. **"Who is beating me, and why?"** A competitor leaderboard showing which businesses AI recommends most and the reasons it gives (reviews, specialties, hours, location).
3. **"What do I do about it?"** A prioritized action plan in plain English (e.g., "Add your Sunday hours to your Google Business Profile and website," "Get listed on Yelp and the local chamber directory, which AI cites often").

### Who uses it?
| User | How they use Findable |
|---|---|
| Small business owner | Runs a monthly audit, follows the action checklist |
| Marketing agency | Monitors many clients, exports white-label reports |
| Franchise marketing manager | Compares visibility across locations |

### Core features
- **Prompt Generator:** auto-creates 50 to 200 realistic customer questions for the business's category and city
- **AI Answer Scanner:** runs those prompts through the OpenAI API (with live web search enabled) to mimic what real customers see
- **Mention Extractor:** pulls out which businesses were named, their rank, sentiment, and cited sources
- **Score + Trends:** tracks visibility over time so owners can see if changes worked
- **Action Plan:** AI-written recommendations tied to the specific gaps found

## 4. How It Works

### Input → OpenAI API → Output

| Stage | What happens | Role of the OpenAI API |
|---|---|---|
| **1. Input** | Owner enters business name, website, category, city, and top competitors (optional) | None |
| **2. Prompt generation** | App creates a list of realistic customer questions | **Text generation:** GPT model writes varied, natural prompts ("cheap oil change near ASU," "who does good tints in Scottsdale?") |
| **3. Answer collection** | Each prompt is sent to the model with web search on | **Responses API + built-in `web_search` tool:** simulates what a customer actually sees, including cited sources |
| **4. Extraction** | Each raw answer is converted into clean data | **Structured Outputs:** model returns JSON that follows our schema (business names, rank, sentiment, reasons, cited URLs) |
| **5. Analysis** | Python aggregates results into scores and competitor rankings | None (pandas / standard Python) |
| **6. Recommendations** | App generates a tailored improvement plan | **Text generation:** GPT summarizes gaps and writes a prioritized, plain-English action list |
| **7. Output** | Dashboard + downloadable PDF report | None |

### Simple flow
```
[Business info]
      │
      ▼
[GPT generates 100 customer-style prompts]
      │
      ▼
[Responses API + web_search answers each prompt]  ──►  raw answers + citations
      │
      ▼
[Structured Outputs → JSON: who was mentioned, rank, sentiment, sources]
      │
      ▼
[Python scoring: visibility %, avg rank, share of voice vs competitors]
      │
      ▼
[GPT writes the action plan]  ──►  Dashboard + PDF report
```

### Example structured output (what the API returns per answer)
```json
{
  "prompt": "best brake repair shop in Tempe AZ",
  "businesses_mentioned": [
    {"name": "Desert Auto Care", "rank": 1, "sentiment": "positive", "reason": "strong reviews, fast service"},
    {"name": "Client Business",  "rank": 3, "sentiment": "neutral",  "reason": "mentioned as an option"}
  ],
  "client_mentioned": true,
  "sources_cited": ["yelp.com", "google.com/maps", "reddit.com"]
}
```

### How this connects to what we've learned
- **Python:** loops through prompts, calls the API, stores results in a pandas DataFrame, calculates scores
- **APIs:** authenticated requests to the OpenAI Responses API, handling JSON responses, rate limits, and errors
- **AI:** prompt engineering, tool use (web search), and structured outputs to turn unstructured text into analyzable data

## 6. Considerations

### Resources needed
| Resource | Details |
|---|---|
| **People** | 1 to 2 Python developers, 1 designer/front-end, 1 marketing/sales lead |
| **Technology** | OpenAI API (Responses API, web search tool, Structured Outputs), Python backend (FastAPI or Flask), database (PostgreSQL), front end (React or Streamlit for MVP), hosting (AWS / Render) |
| **Data** | Business info from the user; optional Google Business Profile or Places API data for enrichment |
| **Budget** | API usage (charged per token plus per web search call), hosting, and initial marketing. Costs scale with the number of prompts per audit, so we would cap prompts per plan tier |
| **Timeline** | ~6 to 8 weeks for an MVP (audit + dashboard), then add tracking and agency features |

### Risks and limitations
| Risk | Why it matters | Mitigation |
|---|---|---|
| **AI answers vary** | The same question can give different answers each time | Run each prompt multiple times and report averages and ranges, not single results |
| **API ≠ consumer ChatGPT** | API results may differ from what users see in the ChatGPT app (personalization, location, memory) | Be transparent: label results as an estimate; use location-specific prompts; validate with spot checks |
| **Only covers one AI platform** | Customers also use Gemini, Perplexity, Copilot, Claude | Start with OpenAI (most-used), add other providers in later phases |
| **API cost** | Hundreds of web-search calls per audit add up | Cache results, use smaller models for extraction, limit prompts by tier |
| **Hallucinations** | The model may invent businesses or misstate details | Cross-check extracted names against Google Places data; flag unverified results |
| **Can't guarantee results** | We can recommend fixes but can't control what AI recommends | Set clear expectations; show trends over time rather than promising rankings |
| **Vendor dependence** | Pricing or policy changes from OpenAI affect the product | Design the code to be model-agnostic so providers can be swapped |
| **Competition** | Larger SEO platforms are adding AI visibility features | Focus on the underserved small-business segment with simple pricing and plain-English advice |
| **Privacy / terms of use** | Must follow OpenAI usage policies and protect customer data | Collect only public business info; follow OpenAI data usage policies |

### Ethical note
The goal is to help businesses be **accurately** represented in AI answers (correct hours, services, reviews), not to manipulate or spam AI systems. Recommendations will focus on legitimate improvements like complete listings, real reviews, and clear website content.
